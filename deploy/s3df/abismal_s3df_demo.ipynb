# abismal on S3DF

Graphical interface for [abismal](https://github.com/rs-station/abismal), a merging program for serial and conventional diffraction data, running on an S3DF GPU node.

Everything comes from a shared, read-only install; nothing needs to be installed. Your copy of this notebook lives in `~/abismal-demo`; results go to `~/abismal-demo/runs`.

1. Run the first cell to check that the session has a GPU.
2. Run the second cell to open the GUI, and copy the example settings it prints into the form.
3. Press **Run** in the GUI. A short example run takes a few minutes on one GPU; refinement against the lysozyme model follows on the node's CPUs.

```python
import torch, abismal, importlib.metadata as md

print("abismal", md.version("abismal"), "| torch", torch.__version__)
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
else:
    print("No GPU in this session. abismal will run on the CPU, very slowly: "
          "relaunch the Jupyter app on a GPU partition with a GPU requested.")
```

```python
import os
from abismal.gui import launch

demo = os.environ.get("ABISMAL_DEMO_DIR", os.path.expanduser("~/abismal-demo"))
hewl = os.path.join(demo, "hewl")
print("Example: room-temperature sulfur-SAD lysozyme, unmerged. Settings to enter:")
print(f"  inputs                 {hewl}/unmerged.mtz")
print(f"  --out-dir              {demo}/runs/hewl")
print( "  --dmin                 1.7      --anomalous")
print(f"  --torchref-pdb         {hewl}/RTSAD_HEWL_refine_25.pdb")
print(f"  --r-free-mtz           {hewl}/r-free-flags.mtz")
print( "  --torchref-wavelength  1.892")
print( "  --epochs               10       (enough to see the sulfur peaks appear)")

launch()
```

**If the GUI shows as plain text** instead of a form, the session started a Jupyter without widget support: in the launcher, check that the custom script ran (its message appears in the session's output log), and pick the *abismal (S3DF)* kernel.

**Your own data:** browse to it with the file pickers in the form. `.mtz` and CrystFEL `.stream` files work as they are; DIALS `.expt`/`.refl` files need the shared install to include DIALS.
