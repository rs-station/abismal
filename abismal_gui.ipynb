# Abismal GUI

Graphical interface for running [abismal](https://github.com/rs-station/abismal) serial diffraction merging.

**JupyterLab / HPC (Open OnDemand):** files are browsed server-side — navigate to your data using the file browser in the GUI.  
**Google Colab:** upload your files to the session first (the Files pane in the left sidebar, or `google.colab.files.upload()`), then point the file browser at `/content`.


```python
# Install abismal if not already present (needed on Google Colab).
# The gui extra is what pulls ipywidgets and solara; without it the next cell
# cannot import anything.
try:
    import abismal.gui
except ImportError:
    import subprocess, sys
    subprocess.check_call([sys.executable, '-m', 'pip', 'install', 'abismal[gui]'])
```

```python
from abismal.gui import ArgparseGUI

gui = ArgparseGUI()
gui.to_widget()
```
