# Abismal GUI

Graphical interface for running [abismal](https://github.com/rs-station/abismal) serial diffraction merging.

**JupyterLab / HPC (Open OnDemand):** Files are browsed server-side — navigate to your data using the file browser in the GUI.  
**Google Colab:** Files are uploaded from your local machine using the upload buttons in the GUI.

```python
# Install abismal if not already present (needed on Google Colab)
try:
    import abismal
except ImportError:
    import subprocess, sys
    subprocess.check_call([sys.executable, '-m', 'pip', 'install', 'abismal'])
```

```python
from abismal.gui import ArgparseGUI

gui = ArgparseGUI()
gui.to_widget()
```

## Results viewer

If you supplied a Phenix `.eff` file, the density map viewer updates automatically during training (above).  
You can also point the viewer at any PDB + MTZ pair after the run:

```python
from abismal.gui import UglyMolViewer

# Adjust paths to your output directory
viewer = UglyMolViewer(
    pdb_file='path/to/structure_refine_001.pdb',
    mtz_file='path/to/structure_refine_001.mtz',
)
viewer.display()
```
